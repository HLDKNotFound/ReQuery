# ReQuery: Dataset Analysis Notebook

This notebook explores the dataset distribution, query/rewrite token statistics, and coreference patterns across **QReCC**, **CANARD**, and **AmbigQA**.

```python
import json
import os
from collections import Counter

# Load processed training split
train_path = '../data/processed/train.jsonl'
records = []
with open(train_path, 'r', encoding='utf-8') as f:
    for line in f:
        if line.strip():
            records.append(json.loads(line))

print(f'Total training records: {len(records):,}')
dataset_counts = Counter(r['dataset'] for r in records)
print('Dataset distribution:', dict(dataset_counts))
```

```python
# Compare query length vs rewrite length
q_lens = [len(r['query'].split()) for r in records]
rw_lens = [len(r['rewrite'].split()) for r in records]

print(f'Mean original query tokens: {sum(q_lens)/len(q_lens):.2f}')
print(f'Mean rewritten query tokens: {sum(rw_lens)/len(rw_lens):.2f}')
print(f'Token expansion ratio: {(sum(rw_lens)/sum(q_lens)):.2f}x')
```

```python
# Display representative samples
for r in records[:5]:
    print('=' * 60)
    print('Dataset:', r['dataset'])
    print('Context:', r.get('context'))
    print('Original Query:', r['query'])
    print('Gold Rewrite:', r['rewrite'])
```
