# ReQuery: Retrieval Evaluation Analysis Notebook

This notebook visualizes and analyzes retrieval metrics, RRF ranking behavior, query diversity, and temperature ablations.

```python
import json
import os

bench_file = '../experiments/benchmark_results.json'
if os.path.exists(bench_file):
    with open(bench_file, 'r') as f:
        data = json.load(f)
    print('Loaded benchmark results!')
    print('Summary Rows:')
    for row in data.get('summary', []):
        print(f"{row['Method']:<22} | Recall@10: {row['Recall@10']:.4f} | MRR@10: {row['MRR@10']:.4f} | Δ MRR: {row['MRR_Improvement_%']:+.2f}%")
else:
    print(f'File {bench_file} not found.')
```

```python
# Multi-Query Diversity analysis
if os.path.exists(bench_file):
    div = data.get('diversity_analysis', {})
    print('Diversity Analysis Across K:')
    for k_val, metrics in div.items():
        print(k_val, metrics)
```

```python
# Fusion Method Comparison
if os.path.exists(bench_file):
    fusion_data = data.get('fusion_ablation', {})
    print('Fusion Ablation (K=3):', fusion_data)
```
